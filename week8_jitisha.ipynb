# 🧠 Single Agent Pipeline Project

## Problem Statement
Build a **Single-Agent Smart Assistant** that:
- Understands user queries
- Routes tasks based on intent
- Uses tools when required
- Returns structured JSON output

### The agent should handle:
- Math queries → Calculator Tool
- Keyword extraction → Keyword Tool
- General queries → Direct response

---
### 🛠️ What You Need to Implement
- Agent logic
- Conditional routing
- Tool integration
- Basic error handling

### 🚀 Bonus

````python
# 🛠️ TOOL 1: Calculator

def calculator(expression: str) -> str:
    """Evaluate a mathematical expression."""
    try:
        return str(eval(expression))
    except Exception:
        return "Error in calculation"
````

````python
# 🛠️ TOOL 2: Keyword Extractor

def extract_keywords(text: str) -> list:
    """Extract keywords from text."""
    try:
        words = text.split()
        keywords = list(set([w.lower() for w in words if len(w) > 4]))
        return keywords[:5]
    except Exception:
        return []
````

## 🤖 Implement Agent Logic Below

👉 Use conditional routing:
- If query contains "calculate" → use calculator
- If query contains "keywords" → use keyword extractor
- Else → general response

````python
# 📝 BONUS: Logging setup

import logging
import re

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)
````

````python
# 🤖 AGENT FUNCTION (IMPLEMENTED)

def agent(query: str):
    """
    Single-agent router:
    - Understands the query
    - Routes to the correct tool based on intent (conditional routing)
    - Handles errors gracefully
    - Returns a structured JSON-like dict
    """
    try:
        if not query or not query.strip():
            return {"type": "error", "result": "Empty query received."}

        query_lower = query.lower()

        # ---- Route 1: Calculator ----
        if "calculate" in query_lower:
            # Pull out the mathematical expression (digits, operators, brackets)
            expression = re.sub(r"calculate", "", query_lower)
            expression = re.sub(r"[^0-9+\-*/().\s]", "", expression).strip()

            if not expression:
                logging.warning(f"No valid expression found in query: {query}")
                return {"type": "error", "result": "No valid mathematical expression found."}

            result = calculator(expression)
            logging.info(f"[Calculator Tool] '{expression}' -> {result}")

            if result == "Error in calculation":
                return {"type": "error", "result": result}
            return {"type": "calculation", "result": result}

        # ---- Route 2: Keyword Extractor ----
        elif "keywords" in query_lower:
            # Extract keywords from the part of the query after 'keywords from'
            text = re.sub(r".*keywords( from)?", "", query, flags=re.IGNORECASE).strip()
            text = text if text else query

            keywords = extract_keywords(text)
            logging.info(f"[Keyword Tool] extracted -> {keywords}")

            if not keywords:
                return {"type": "error", "result": "No keywords could be extracted."}
            return {"type": "keywords", "result": keywords}

        # ---- Route 3: General Response ----
        else:
            logging.info(f"[General Response] query -> {query}")
            return {
                "type": "general",
                "result": f"I received your query: '{query}'. This doesn't need a specific tool, so here's a direct response."
            }

    except Exception as e:
        logging.error(f"Agent encountered an error: {e}")
        return {"type": "error", "result": f"Something went wrong: {str(e)}"}
````

## 📦 Expected Output Format

```
{
  "type": "calculation / keywords / general / error",
  "result": ...
}
```

````python
# 🧪 Test Cases

queries = [
    "Calculate 20 + 5",
    "Extract keywords from Artificial Intelligence is transforming industries",
    "What is machine learning?"
]

for q in queries:
    print("Query:", q)
    print("Response:", agent(q))
    print("-" * 50)
````
